```python
import pandas as pd
import sqlite3

customers = pd.read_csv("../data/customers.csv")
conn = sqlite3.connect("../data/retail.db")
customers.to_sql("customers", conn, if_exists="replace", index=False)

query = """
SELECT
    CASE WHEN frequency = 1 THEN '1 order'
         WHEN frequency BETWEEN 2 AND 3 THEN '2-3 orders'
         WHEN frequency BETWEEN 4 AND 9 THEN '4-9 orders'
         ELSE '10+ orders' END AS frequency_group,
    COUNT(*) AS customers,
    ROUND(AVG(buys_again), 3) AS buy_again_rate,
    ROUND(AVG(future_spend), 1) AS avg_future_spend
FROM customers
GROUP BY frequency_group
ORDER BY buy_again_rate
"""

print(pd.read_sql(query, conn))
```
